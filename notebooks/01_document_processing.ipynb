```python
from src.document.parser import DocumentParser
from src.document.cleaner import DocumentCleaner
from src.document.chunker import DocumentChunker
from src.document.metadata import DocumentMetadataBuilder
```

```python
file_path = r'C:\Users\Mohammed Gamal\OneDrive\Desktop\DEPI\AI-Educational-Video-Generator\data\sample_documents\deep_learning_guide.docx'

parser = DocumentParser.get_parser(file_path)
cleaner = DocumentCleaner()
chunker = DocumentChunker()
metadata_builder = DocumentMetadataBuilder()
```

```python
# Step 1: Parse
parsed_data = parser.parse(file_path)

# Step 2: Clean
cleaned_data = cleaner.clean(file_path, parsed_data)

# Step 3: Chunk
chunks = chunker.chunk(cleaned_data )

# Step 4: Build metadata
documents = metadata_builder.build(file_path, chunks)
```

```python
print("Parsed elements:", len(parsed_data))
print("Cleaned elements:", len(cleaned_data))
print("Chunks:", len(chunks))
print("Final documents:", len(documents))

for document in documents[:3]:
    print("\n" + "=" * 60)
    print("Text:")
    print(document["text"])

    print("\nMetadata:")
    print(document["metadata"])
```
